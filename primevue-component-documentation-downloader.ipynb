```python
import os
import requests

BASE_URL = "https://primevue.org/llms/components/"
OUTPUT_FOLDER = "primevue/component-documentation/raw"

COMPONENTS = [
    # --- level 1 (15 PVK) ---

    # forms
    "checkbox",
    "inputnumber",
    "inputtext",
    "password",
    "radiobutton",
    "slider",
    "textarea",
    "toggleswitch",

    # buttons
    "button",

    # panel
    "divider",

    # misc
    "avatar",
    "badge",
    "progressbar",
    "skeleton",
    "tag",


    # --- level 2 (5 PVK) ---

    # panel
    "accordion",
    "card",
    "tabs",

    # menu
    'breadcrumb',
    'menu',


    # --- level 3 (5 PVK) ---

    # form
    "datepicker",
    "select",

    # figma-data
    "datatable",

    # overlay
    'dialog',
    'popover'
]

def download_markdowns(components, base_url, output_folder):
    print(f"\n Starting download of {len(components)} files...\n")

    success_count = 0
    error_count = 0

    for component in components:
        file_name = f"{component}.md"
        url = f"{base_url}{file_name}"
        file_path = os.path.join(output_folder, file_name)

        try:
            response = requests.get(url, timeout=10)

            if response.status_code == 200:
                with open(file_path, 'w', encoding='utf-8') as f:
                    f.write(response.text)

                print(f"✅ [{component}] successfully saved: {file_name}")

                success_count += 1
            elif response.status_code == 404:
                print(f"❌ [{component}] not found (404 Error) at {url}")

                error_count += 1
            else:
                print(f"⚠️ [{component}] Error: HTTP {response.status_code}")
                error_count += 1

        except requests.exceptions.RequestException as e:
            print(f"💥 [{component}] Network error: {e}")
            error_count += 1

    print("Download finished:")
    print(f"   Successfully: {success_count}")
    print(f"   Failed: {error_count}")

download_markdowns(COMPONENTS, BASE_URL, OUTPUT_FOLDER)
```
