```python
import os
import json
import requests
```

```python
FIGMA_TOKEN = "figd_FllSRYcsg46AlvwsUr9dFMYq0el9eWAo37Ee3Yuc"
FILE_KEY = "2amUWk9ELJwZKqVTmIBIIv"
NODE_ID = "10:125"

HEADERS = {
    "X-Figma-Token": FIGMA_TOKEN
}

BASE_URL = "https://api.figma.com/v1"

FOLDER_PATH = "raw-data"
COMPONENT_NAME = "button"
```

```python
def get_node_figma_file(file_key, node_id):
    url = f"{BASE_URL}/files/{file_key}/nodes?ids={node_id}"

    print(f"Send request to: {url} ...")

    response = requests.get(url, headers=HEADERS)

    if response.status_code == 200:
        print("Successfully retrieved data from Figma API")

        return response.json()
    else:
        print(f"Error {response.status_code}: {response.text} - {response.headers}")

        return None

figma_data = get_node_figma_file(FILE_KEY, NODE_ID)
```

```python
if figma_data:
    if not os.path.exists(FOLDER_PATH):
        os.makedirs(FOLDER_PATH)

    file_path = os.path.join(FOLDER_PATH, f"{COMPONENT_NAME}-component-variants.json")

    with open(file_path, "w", encoding="utf-8") as json_file:
        json.dump(figma_data, json_file, indent=4, ensure_ascii=False)

    print(f"Data successfully saved to {file_path}")
else:
    print("No data to save")
```
