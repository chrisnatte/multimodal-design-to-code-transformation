```python
def get_node_figma_file(file_key, node_id):
    url = f"{BASE_URL}/files/{file_key}/nodes?ids={node_id}"

    print(f"Send request to: {url} ...")

    response = requests.get(url, headers=HEADERS)

    if response.status_code == 200:
        print("Successfully retrieved figma-data from Figma API")

        return response.json()
    else:
        print(f"Error {response.status_code}: {response.text} - {response.headers}")

        return None
```

```python
def save_node_figma_file(name, data):
    if data:
        if not os.path.exists(FOLDER_PATH):
            os.makedirs(FOLDER_PATH)

        file_path = os.path.join(FOLDER_PATH, f"{name}.json")

        with open(file_path, "w", encoding="utf-8") as json_file:
            json.dump(data, json_file, indent=4, ensure_ascii=False)

        print(f"Data successfully saved to {file_path}")
    else:
        print("No figma-data to save")
```

```python
for component in COMPONENTS_VARIANTS:
    name = component["name"]
    node_id = component["node_id"]

    figma_data = get_node_figma_file(FILE_KEY, node_id)

    save_node_figma_file(name, figma_data)

    time.sleep(30)
```
